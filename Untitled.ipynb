# Kernel Density Estimation with Cross-Validated Bandwidth Selection

**Self-Initiated Statistical / Machine Learning Project**

This project implements Kernel Density Estimation (KDE) from scratch to estimate an unknown probability density function from observed data.

The bandwidth parameter is selected using leave-one-out cross-validation and log joint likelihood. The final estimated density is compared with the known true probability density function of a synthetic Gaussian mixture distribution.

## 1. Objective

The objective of this project is to develop a non-parametric probability density estimation framework using Kernel Density Estimation.

The project focuses on:

- Implementing Gaussian Kernel Density Estimation from first principles
- Studying the influence of the bandwidth parameter on density estimation
- Selecting the optimal bandwidth using leave-one-out cross-validation
- Maximizing the log joint likelihood for data-driven parameter selection
- Comparing the estimated probability density with the known true density
- Quantifying the estimation error using MSE and MAE

## 2. Mathematical Formulation

For observations $x_1,x_2,\ldots,x_n$, the Kernel Density Estimator is defined as:

$$
\hat{f}_h(x)
=
\frac{1}{nh}
\sum_{i=1}^{n}
K\left(\frac{x-x_i}{h}\right)
$$

where:

- $n$ is the number of observations
- $h$ is the bandwidth
- $K(\cdot)$ is the kernel function

A Gaussian kernel is used:

$$
K(u)
=
\frac{1}{\sqrt{2\pi}}
e^{-u^2/2}
$$

The bandwidth controls the smoothness of the estimated density. A very small bandwidth can result in an overly noisy estimate, while a very large bandwidth can oversmooth important features of the underlying distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

print("Libraries imported successfully.")
```

## 3. Generate Synthetic Dataset

A two-component Gaussian mixture distribution is used to generate the observations.

Using a known underlying distribution allows the estimated KDE to be compared directly against the true probability density function.

```python
# Number of observations
n = 500

# Parameters of the first Gaussian component
mu1 = -2.0
sigma1 = 0.8

# Parameters of the second Gaussian component
mu2 = 2.0
sigma2 = 1.0

# Probability of selecting the first component
mixing_probability = 0.6

# Randomly assign observations to the two components
component = np.random.rand(n) < mixing_probability

# Generate observations
data = np.where(
    component,
    np.random.normal(mu1, sigma1, n),
    np.random.normal(mu2, sigma2, n)
)

data = data.reshape(-1, 1)

print(f"Number of samples: {n}")
print(f"Sample mean: {data.mean():.3f}")
print(f"Sample standard deviation: {data.std():.3f}")
```

```python
# Visualize the generated dataset

plt.figure(figsize=(9, 5))

plt.hist(
    data.ravel(),
    bins=30,
    density=True,
    alpha=0.7,
    edgecolor="black"
)

plt.xlabel("x")
plt.ylabel("Probability Density")
plt.title("Generated Synthetic Dataset")
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

## 4. True Probability Density Function

Since the synthetic dataset was generated from a known Gaussian mixture, its true PDF can be calculated analytically.

The true distribution is:

$$
f(x)
=
0.6\,\mathcal{N}(-2,0.8^2)
+
0.4\,\mathcal{N}(2,1.0^2)
$$

This true PDF will be used as the reference against which the KDE estimate is evaluated.

```python
def normal_pdf(x, mu, sigma):
    """
    Probability density function of a normal distribution.
    """
    return (
        1 / (sigma * np.sqrt(2 * np.pi))
        * np.exp(-0.5 * ((x - mu) / sigma) ** 2)
    )


# Evaluation points for the PDF
x_grid = np.linspace(
    data.min() - 1,
    data.max() + 1,
    600
)

# Calculate true PDF
true_pdf = (
    mixing_probability * normal_pdf(x_grid, mu1, sigma1)
    + (1 - mixing_probability) * normal_pdf(
        x_grid, mu2, sigma2
    )
)

# Plot true PDF
plt.figure(figsize=(9, 5))

plt.plot(
    x_grid,
    true_pdf,
    linewidth=2,
    label="True PDF"
)

plt.xlabel("x")
plt.ylabel("Probability Density")
plt.title("True Probability Density Function")
plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

## 5. Gaussian Kernel Density Estimator

The KDE is implemented directly from the mathematical formulation rather than using a built-in density estimation function.

For a Gaussian kernel:

$$
\hat{f}_h(x)
=
\frac{1}{nh}
\sum_{i=1}^{n}
\frac{1}{\sqrt{2\pi}}
\exp
\left[
-\frac{1}{2}
\left(
\frac{x-x_i}{h}
\right)^2
\right]
$$

```python
def gaussian_kde(data, x_grid, bandwidth):
    """
    Estimate a one-dimensional probability density
    using Gaussian Kernel Density Estimation.
    """

    samples = np.asarray(data).ravel()
    x_grid = np.asarray(x_grid)

    n_samples = len(samples)

    if bandwidth <= 0:
        raise ValueError("Bandwidth must be positive.")

    # Normalize distance by bandwidth
    u = (
        x_grid[:, None] - samples[None, :]
    ) / bandwidth

    # Gaussian kernel
    kernel_values = (
        np.exp(-0.5 * u**2)
        / np.sqrt(2 * np.pi)
    )

    # KDE equation
    density = (
        kernel_values.sum(axis=1)
        / (n_samples * bandwidth)
    )

    return density
```

```python
# Test KDE with an initial bandwidth

initial_bandwidth = 0.5

initial_estimate = gaussian_kde(
    data,
    x_grid,
    initial_bandwidth
)

plt.figure(figsize=(9, 5))

plt.plot(
    x_grid,
    true_pdf,
    linewidth=2,
    label="True PDF"
)

plt.plot(
    x_grid,
    initial_estimate,
    "--",
    linewidth=2,
    label=f"KDE, h = {initial_bandwidth}"
)

plt.xlabel("x")
plt.ylabel("Probability Density")
plt.title("KDE Estimate vs True PDF")

plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

## 6. Effect of Bandwidth

The bandwidth is one of the most important parameters in KDE.

- A **small bandwidth** produces a highly detailed but potentially noisy estimate.
- A **large bandwidth** produces a smoother but potentially oversimplified estimate.
- An appropriate bandwidth provides a balance between variance and bias.

The following experiment compares KDE estimates for different bandwidth values.

```python
bandwidths = [0.2, 0.5, 1.0]

plt.figure(figsize=(9, 5))

for h in bandwidths:

    estimate = gaussian_kde(
        data,
        x_grid,
        h
    )

    plt.plot(
        x_grid,
        estimate,
        linewidth=1.7,
        label=f"h = {h}"
    )

# True distribution
plt.plot(
    x_grid,
    true_pdf,
    "--",
    linewidth=2,
    label="True PDF"
)

plt.xlabel("x")
plt.ylabel("Probability Density")
plt.title("Effect of Bandwidth on KDE")

plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

## 7. Leave-One-Out Cross-Validation

To avoid selecting the bandwidth manually, leave-one-out cross-validation is used.

For each observation $x_i$, the density at $x_i$ is estimated using all observations except $x_i$ itself.

The resulting log joint likelihood is:

$$
L(h)
=
\sum_{i=1}^{n}
\log
\left[
\hat{f}_{-i,h}(x_i)
\right]
$$

The bandwidth that maximizes $L(h)$ is selected as the optimal bandwidth.

```python
def loo_log_likelihood(data, bandwidth):
    """
    Calculate leave-one-out log likelihood
    for a given KDE bandwidth.
    """

    samples = np.asarray(data).ravel()

    n_samples = len(samples)

    if bandwidth <= 0:
        raise ValueError("Bandwidth must be positive.")

    # Pairwise differences
    differences = (
        samples[:, None]
        - samples[None, :]
    ) / bandwidth

    # Gaussian kernel matrix
    kernel_matrix = (
        np.exp(-0.5 * differences**2)
        / np.sqrt(2 * np.pi)
    )

    # Remove each observation's own contribution
    np.fill_diagonal(
        kernel_matrix,
        0.0
    )

    # Leave-one-out density
    loo_density = (
        kernel_matrix.sum(axis=1)
        / ((n_samples - 1) * bandwidth)
    )

    # Avoid log(0)
    loo_density = np.maximum(
        loo_density,
        1e-12
    )

    # Log joint likelihood
    log_likelihood = np.sum(
        np.log(loo_density)
    )

    return log_likelihood
```

## 8. Bandwidth Optimization

A range of candidate bandwidth values is evaluated.

For each bandwidth, the leave-one-out log joint likelihood is calculated. The bandwidth corresponding to the maximum likelihood is selected as the optimal value.

```python
# Candidate bandwidth values

candidate_bandwidths = np.linspace(
    0.1,
    1.5,
    80
)

# Calculate log likelihood for every bandwidth

log_likelihoods = np.array([
    loo_log_likelihood(
        data,
        h
    )
    for h in candidate_bandwidths
])

# Find maximum likelihood

best_index = np.argmax(
    log_likelihoods
)

optimal_bandwidth = (
    candidate_bandwidths[best_index]
)

maximum_log_likelihood = (
    log_likelihoods[best_index]
)

print(
    f"Optimal bandwidth: "
    f"{optimal_bandwidth:.4f}"
)

print(
    f"Maximum LOO log likelihood: "
    f"{maximum_log_likelihood:.4f}"
)
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    candidate_bandwidths,
    log_likelihoods,
    linewidth=2
)

plt.scatter(
    optimal_bandwidth,
    maximum_log_likelihood,
    s=70,
    label=(
        f"Optimal h = "
        f"{optimal_bandwidth:.3f}"
    )
)

plt.xlabel("Bandwidth")
plt.ylabel(
    "Leave-One-Out Log Likelihood"
)

plt.title(
    "Cross-Validation for "
    "Bandwidth Selection"
)

plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

## 9. Final KDE Using the Optimized Bandwidth

The optimal bandwidth obtained through cross-validation is now used to calculate the final KDE estimate.

```python
# Calculate final KDE

final_estimate = gaussian_kde(
    data,
    x_grid,
    optimal_bandwidth
)

# Plot final result

plt.figure(figsize=(9, 5))

plt.plot(
    x_grid,
    true_pdf,
    linewidth=2,
    label="True PDF"
)

plt.plot(
    x_grid,
    final_estimate,
    "--",
    linewidth=2,
    label=(
        f"KDE, optimized h = "
        f"{optimal_bandwidth:.3f}"
    )
)

plt.xlabel("x")
plt.ylabel("Probability Density")

plt.title(
    "Final KDE Using "
    "Cross-Validated Bandwidth"
)

plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

## 10. Quantitative Evaluation

To quantitatively evaluate the KDE estimate, the estimated PDF is compared with the known true PDF.

Two error metrics are calculated:

### Mean Squared Error

$$
MSE
=
\frac{1}{N}
\sum_{j=1}^{N}
\left(
\hat f(x_j)-f(x_j)
\right)^2
$$

### Mean Absolute Error

$$
MAE
=
\frac{1}{N}
\sum_{j=1}^{N}
\left|
\hat f(x_j)-f(x_j)
\right|
$$

```python
# Calculate evaluation metrics

mse = np.mean(
    (final_estimate - true_pdf) ** 2
)

mae = np.mean(
    np.abs(
        final_estimate - true_pdf
    )
)

print("=" * 45)
print("FINAL KDE RESULTS")
print("=" * 45)

print(
    f"Optimal bandwidth : "
    f"{optimal_bandwidth:.4f}"
)

print(
    f"Maximum log likelihood : "
    f"{maximum_log_likelihood:.4f}"
)

print(
    f"Mean Squared Error : "
    f"{mse:.6f}"
)

print(
    f"Mean Absolute Error : "
    f"{mae:.6f}"
)

print("=" * 45)
```

## 11. Final Comparison

The final KDE estimate obtained using cross-validated bandwidth selection is compared with the true probability density.

The optimized bandwidth provides a data-driven balance between under-smoothing and over-smoothing, producing an estimate that captures the major features of the underlying distribution.

```python
plt.figure(figsize=(10, 6))

plt.plot(
    x_grid,
    true_pdf,
    linewidth=2.5,
    label="True PDF"
)

plt.plot(
    x_grid,
    final_estimate,
    "--",
    linewidth=2.5,
    label="Optimized KDE"
)

plt.fill_between(
    x_grid,
    true_pdf,
    final_estimate,
    alpha=0.15
)

plt.xlabel(
    "x",
    fontsize=12
)

plt.ylabel(
    "Probability Density",
    fontsize=12
)

plt.title(
    "True PDF vs Cross-Validated KDE",
    fontsize=14
)

plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```
